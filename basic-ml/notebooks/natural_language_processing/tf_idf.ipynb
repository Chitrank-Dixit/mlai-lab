```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# 1. Prepare a corpus where certain words carry contextual dependencies
corpus = [
    "The data science pipeline is highly efficient.",
    "This predictive machine learning model is not efficient.",
    "Data science requires robust cloud infrastructure infrastructure."
]
```

```python
# 2. Initialize the TF-IDF Vectorizer
# - stop_words='english': Removes generic filler words (the, is, a)
# - ngram_range=(1, 2): Captures both single words (unigrams) and word pairs (bigrams)
tfidf_vectorizer = TfidfVectorizer(stop_words='english', ngram_range=(1, 2))
```

```python
# 3. Transform the text into a sparse TF-IDF feature matrix
tfidf_matrix = tfidf_vectorizer.fit_transform(corpus)
```

```python
# 4. Convert to a structured DataFrame for inspection
feature_names = tfidf_vectorizer.get_feature_names_out()
tfidf_df = pd.DataFrame(data=tfidf_matrix.toarray(), columns=feature_names)

print("--- TF-IDF Matrix Extraction (Sample Columns) ---")
# Displaying a selection of interesting unigrams and bigrams
target_cols = ['data', 'data science', 'efficient', 'model efficient', 'infrastructure']
print(tfidf_df[target_cols].round(4))
```
