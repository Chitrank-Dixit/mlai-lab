```python
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
# 1. Synthesize a dataset with 4 features (highly redundant)
np.random.seed(42)
x1 = np.random.normal(0, 1, 100)
x2 = x1 * 2 + np.random.normal(0, 0.1, 100)  # Heavily correlated with x1
x3 = np.random.normal(5, 2, 100)
x4 = x3 * -0.5 + np.random.normal(0, 0.2, 100) # Heavily correlated with x3

df = pd.DataFrame({'Feat1': x1, 'Feat2': x2, 'Feat3': x3, 'Feat4': x4})
```

```python
# 2. PCA MANDATE: Standardize features before applying the transformation
scaled_data = StandardScaler().fit_transform(df)
```

```python
# 3. Initialize PCA to compress from 4 dimensions down to 2
pca = PCA(n_components=2)
principal_components = pca.fit_transform(scaled_data)
```

```python
# 4. Convert to a clean DataFrame
pca_df = pd.DataFrame(data=principal_components, columns=['PC1', 'PC2'])
```

```python
print("--- Compressed Dataset Shape ---")
print(f"Original Shape:   {df.shape}")
print(f"Compressed Shape: {pca_df.shape}\n")

print("--- Information Retention Breakdown ---")
variance_ratios = pca.explained_variance_ratio_
print(f"Variance explained by PC1: {variance_ratios[0]*100:.2f}%")
print(f"Variance explained by PC2: {variance_ratios[1]*100:.2f}%")
print(f"Total Information Retained: {np.sum(variance_ratios)*100:.2f}%")
```
