```python
import numpy as np
from sklearn.linear_model import LogisticRegression
```

```python
# 1. Prepare Dummy Data
# X = [Number of misspelled words, Links inside email]
X = np.array([[0, 1], [1, 0], [5, 4], [8, 3], [0, 0], [6, 5]])
# y = 0 (Not Spam), 1 (Spam)
y = np.array([0, 0, 1, 1, 0, 1])

```

```python
# 2. Initialize and train the classifier
classifier = LogisticRegression()
classifier.fit(X, y)

```

```python
# 3. Predict the hard class for a new email 
# (e.g., 4 misspelled words, 2 links)
new_email = np.array([[4, 2]])
prediction = classifier.predict(new_email)
probability = classifier.predict_proba(new_email)

print(f"Predicted Class:       {prediction[0]} (1 = Spam, 0 = Safe)")
print(f"Probability of Safe:  {probability[0][0]:.4f}")
print(f"Probability of Spam:  {probability[0][1]:.4f}")
```
