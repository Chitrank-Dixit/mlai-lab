```python
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report
```

```python
# Actual true labels vs. The predictions generated by a model
y_true = np.array([0, 0, 1, 1, 0, 1, 0, 0, 1, 0])  # 1 = Spam, 0 = Safe
y_pred = np.array([0, 0, 1, 0, 0, 1, 1, 0, 1, 0])  # The model missed one spam, flagged one safe
```

```python
# 1. Generate the raw Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
tn, fp, fn, tp = cm.ravel()

print("--- Confusion Matrix Breakdown ---")
print(f"True Negatives (TN):  {tn}")
print(f"False Positives (FP): {fp}  <-- Type I Error")
print(f"False Negatives (FN): {fn}  <-- Type II Error")
print(f"True Positives (TP):  {tp}\n")

print("--- Comprehensive Classification Report ---")
# 2. Generate precision, recall, and f1-score automatically
print(classification_report(y_true, y_pred, target_names=['Safe (0)', 'Spam (1)']))
```
