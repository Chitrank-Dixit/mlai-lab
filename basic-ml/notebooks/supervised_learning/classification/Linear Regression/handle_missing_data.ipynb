```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
```

```python
# 1. Create a messy raw dataset
data = pd.DataFrame({
    'Age': [25, np.nan, 47, 35, 22],
    'Income': [50000, 120000, 85000, np.nan, 42000],
    'City': ['Indore', 'Mumbai', 'Indore', 'Delhi', 'Mumbai']
})
```

```python
# 2. Impute missing numerical values using the median
imputer = SimpleImputer(strategy='median')
data[['Age', 'Income']] = imputer.fit_transform(data[['Age', 'Income']])
```

```python
# 3. Scale numerical features (Standardization)
scaler = StandardScaler()
data[['Age', 'Income']] = scaler.fit_transform(data[['Age', 'Income']])
```

```python
# 4. One-Hot Encode categorical features
encoder = OneHotEncoder(sparse_output=False)
encoded_cities = encoder.fit_transform(data[['City']])
city_columns = encoder.get_feature_names_out(['City'])
encoded_city_df = pd.DataFrame(encoded_cities, columns=city_columns)
```

```python
# Combine into a final, ML-ready DataFrame
final_dataset = pd.concat([data.drop(columns=['City']), encoded_city_df], axis=1)
print(final_dataset)
```
