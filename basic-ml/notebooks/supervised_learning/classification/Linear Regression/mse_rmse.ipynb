```python
import numpy as np
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Actual true prices vs. Predictions made by the model
y_true = np.array([200, 290, 340, 410, 470, 550])
y_pred = np.array([202, 285, 344, 408, 475, 546]) # Hypothesized outputs from our line
```

```python
# 1. Calculate MSE, then take the square root for RMSE
mse = mean_squared_error(y_true, y_pred)
rmse = np.sqrt(mse)
```

```python
# 2. Calculate R2 Score
r2 = r2_score(y_true, y_pred)

print(f"Root Mean Squared Error (RMSE): ${rmse:.2f}k")
print(f"R-squared (R2) Score:           {r2:.4f} ({r2*100:.1f}% variance explained)")
```
