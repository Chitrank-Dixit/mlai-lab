```python
import pandas as pd
import numpy as np
```

```python

# 1. Synthesize a dataset with redundant features
data = pd.DataFrame({
    'House_Size': [1000, 1500, 1800, 2200, 2500],
    'Num_Bedrooms': [2, 3, 3, 4, 4],            # Highly correlated with House_Size
    'Distance_to_School': [0.5, 1.2, 3.4, 0.8, 2.1],
    'Lot_Size': [4000, 5000, 6200, 7000, 8500],
    'Target_Price': [200, 290, 340, 410, 470]
})

```

```python
# 2. Calculate the absolute correlation matrix
corr_matrix = data.drop(columns=['Target_Price']).corr().abs()

```

```python
# 3. Select the upper triangle of the correlation matrix 
# (Since it's symmetrical, we only look at the top half above the diagonal)
upper_tri = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

```

```python
# 4. Find features with a correlation greater than a threshold (e.g., 0.90)
to_drop = [column for column in upper_tri.columns if any(upper_tri[column] > 0.90)]

print(f"Features flagged for removal (Redundant): {to_drop}")

```

```python
# 5. Drop the redundant features
clean_data = data.drop(columns=to_drop)
print("\nCleaned Dataset:")
print(clean_data)
```
