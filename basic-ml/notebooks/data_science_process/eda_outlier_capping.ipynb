```python
import numpy as np
import pandas as pd
```

```python
# 1. Synthesize a dataset with normal values and extreme outliers
np.random.seed(42)
salaries = np.random.normal(50000, 10000, 95).tolist()
outliers = [250000, 300000, 12000, 8000, 5000] # Extreme anomalies
data = pd.DataFrame({'Salary': salaries + outliers})

print("--- Step 1: Initial Raw Statistics ---")
# describe() automatically extracts the 5-number summary needed for EDA
print(data.describe().round(2))
```

```python
# 2. Compute Q1, Q3, and IQR
Q1 = data['Salary'].quantile(0.25)
Q3 = data['Salary'].quantile(0.75)
IQR = Q3 - Q1
```

```python
# 3. Establish mathematical fences
lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

print(f"\n--- Step 2: Outlier Boundaries ---")
print(f"Lower Fence: ${lower_fence:.2f}")
print(f"Upper Fence: ${upper_fence:.2f}")
```

```python
# 4. Identify the anomalies
flagged_outliers = data[(data['Salary'] < lower_fence) | (data['Salary'] > upper_fence)]
print(f"Number of Outliers Detected: {len(flagged_outliers)}")
```

```python
# 5. Handle Outliers via Capping (Winsorization)
data['Salary_Cleaned'] = np.clip(data['Salary'], lower_fence, upper_fence)

print(f"\n--- Step 3: Cleaned Statistics (Post-Capping) ---")
print(data['Salary_Cleaned'].describe().round(2))
```
