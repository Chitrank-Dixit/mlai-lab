```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.preprocessing import StandardScaler
```

```python
# 1. Synthesize data: 2 true signals, 2 pure random noise columns
np.random.seed(42)
X1 = np.random.normal(0, 1, 50)
X2 = np.random.normal(0, 1, 50)
Noise1 = np.random.normal(0, 1, 50)
Noise2 = np.random.normal(0, 1, 50)

# True Target depends strictly on X1 and X2
y = (3 * X1) + (5 * X2) + np.random.normal(0, 0.5, 50)
X = pd.DataFrame({'True_X1': X1, 'True_X2': X2, 'Noise_1': Noise1, 'Noise_2': Noise2})

# REGULARIZATION MANDATE: You must scale features before regularizing. 
# Otherwise, the penalty treats features differently based on their scale.
X_scaled = StandardScaler().fit_transform(X)
```

```python
# 2. Train unregularized vs regularized models
# alpha=1.0 sets a moderate penalty scale
lr = LinearRegression().fit(X_scaled, y)
ridge = Ridge(alpha=1.0).fit(X_scaled, y)
lasso = Lasso(alpha=0.5).fit(X_scaled, y)
```

```python
# 3. Compile coefficients for audit
coef_comparison = pd.DataFrame({
    'Linear Regression': lr.coef_,
    'Ridge (L2)': ridge.coef_,
    'Lasso (L1)': lasso.coef_
}, index=X.columns).round(4)

print("--- Coefficient Regularization Impact ---")
print(coef_comparison)
```
