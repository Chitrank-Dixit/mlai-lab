```python
import numpy as np
import pandas as pd
from statsmodels.tsa.stattools import adfuller
from statsmodels.tsa.arima.model import ARIMA
```

```python
# 1. Synthesize a non-stationary time series (Random walk with an upward trend)
np.random.seed(42)
n_steps = 100
time_series = np.cumsum(np.random.normal(loc=0.5, scale=2, size=n_steps))
dates = pd.date_range(start='2026-01-01', periods=n_steps, freq='D')
df = pd.Series(time_series, index=dates)
```

```python
# 2. Check Stationarity via the Augmented Dickey-Fuller (ADF) Test
adf_result = adfuller(df)
print("--- Step 1: Initial Stationarity Audit ---")
print(f"ADF Statistic: {adf_result[0]:.4f}")
print(f"p-value:       {adf_result[1]:.4f} (If > 0.05, data is non-stationary)\n")
```

```python
# 3. Fit an ARIMA(1, 1, 1) Model
# We set d=1 because the data has a clear trend and needs 1 round of differencing
model = ARIMA(df, order=(1, 1, 1))
results = model.fit()

print("--- Step 2: ARIMA Model Summary Extract ---")
print(results.summary().tables[1])
```

```python
# 4. Forecast the next 5 days out into the future
forecast = results.forecast(steps=5)
print("\n--- Step 3: Future 5-Day Forecast ---")
print(forecast.round(4))
```
