```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures

# Create a small sample dataframe
df = pd.DataFrame({
    'Velocity': [10, 20, 30],
    'Mass': [1000, 1200, 1500]
})

# Initialize the polynomial transformer for degree 2, interaction terms included
poly = PolynomialFeatures(degree=2, include_bias=False)
poly_features = poly.fit_transform(df)

# Convert to a clean DataFrame to see the engineered columns
poly_cols = poly.get_feature_names_out(['Velocity', 'Mass'])
poly_df = pd.DataFrame(poly_features, columns=poly_cols)

print("--- Automated Feature Expansion ---")
print(poly_df)
```
