```python
import pandas as pd
from sklearn.datasets import load_digits
from sklearn.manifold import TSNE
```

```python
# 1. Load the Digits dataset (8x8 pixel images of handwritten digits 0-9)
# Each sample has 64 numerical features (pixel intensities)
digits = load_digits()
X = digits.data
y = digits.target

print(f"Original high-dimensional dataset shape: {X.shape}")
```

```python
# 2. Initialize and fit t-SNE
# - n_components=2: Project onto a 2D plane for visualization
# - perplexity=30: Standard operational neighborhood window
# - max_iter=1000: Total optimization iterations for KL divergence convergence (Renamed from n_iter in v1.5)
tsne = TSNE(n_components=2, perplexity=30, max_iter=1000, random_state=42)
X_embedded = tsne.fit_transform(X)
```

```python
# 3. Compile the 2D coordinates into a structured DataFrame
tsne_df = pd.DataFrame(data=X_embedded, columns=['Dimension_1', 'Dimension_2'])
tsne_df['Digit_Label'] = y

print(f"Compressed low-dimensional manifold shape: {tsne_df.shape}\n")
print("--- First 5 rows of mapped coordinates ---")
print(tsne_df.head())
```
