```python
import numpy as np
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
from sklearn.metrics import silhouette_score, davies_bouldin_score
```

```python
# 1. Generate a synthetic dataset with 3 distinct isotropic Gaussian blobs
X, _ = make_blobs(n_samples=500, n_features=2, centers=3, cluster_std=0.60, random_state=42)
```

```python
# 2. Run K-Means with the correct number of clusters (K=3)
kmeans_correct = KMeans(n_clusters=3, random_state=42, n_init='auto').fit(X)
labels_correct = kmeans_correct.labels_
```

```python
# 3. Run K-Means with a poorly chosen number of clusters (K=6)
kmeans_poor = KMeans(n_clusters=6, random_state=42, n_init='auto').fit(X)
labels_poor = kmeans_poor.labels_
```

```python
# 4. Evaluate and compare the models
print("--- Model Validation Metrics (K=3 vs K=6) ---")
print(f"[Optimal K=3] Silhouette Score:       {silhouette_score(X, labels_correct):.4f}  (Higher is better)")
print(f"[Optimal K=3] Davies-Bouldin Index:   {davies_bouldin_score(X, labels_correct):.4f}  (Lower is better)\n")

print(f"[Poor K=6]    Silhouette Score:       {silhouette_score(X, labels_poor):.4f}")
print(f"[Poor K=6]    Davies-Bouldin Index:   {davies_bouldin_score(X, labels_poor):.4f}")
```
