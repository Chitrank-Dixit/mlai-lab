```python
import numpy as np
import torch
import torch.nn as nn

# Same XOR data as notebook 1 (self-contained)
X = np.array([[0., 0.],
              [0., 1.],
              [1., 0.],
              [1., 1.]])
y = np.array([[0.],
              [1.],
              [1.],
              [0.]])

X_t = torch.tensor(X, dtype=torch.float32)
y_t = torch.tensor(y, dtype=torch.float32)

model = nn.Sequential(
    nn.Linear(2, 4),
    nn.Sigmoid(),
    nn.Linear(4, 1),
    nn.Sigmoid(),
)
opt = torch.optim.SGD(model.parameters(), lr=1.0)
loss_fn = nn.MSELoss()

for epoch in range(2000):
    opt.zero_grad()
    pred = model(X_t)
    loss = loss_fn(pred, y_t)
    loss.backward()
    opt.step()
    if epoch % 200 == 0:
        print(epoch, float(loss))

with torch.no_grad():
    print("predictions:", model(X_t).numpy().ravel())
```
