```python
import numpy as np

# 1. Setup sample input data (X) and true target label (y)
X = np.array([2.0, 3.0])
y_true = 12.0
```

```python
# 2. Initialize weights and bias randomly
weights = np.array([0.5, 0.5])
bias = 0.0
learning_rate = 0.01

print("--- Initial System State ---")
print(f"Initial Weights: {weights}, Bias: {bias}")

# =====================================================================
# STEP 1: FORWARD PROPAGATION
# =====================================================================
# Using a linear neuron (no activation function for basic regression demo)
y_pred = np.dot(X, weights) + bias
print(f"Prediction (y_pred): {y_pred:.4f}")

# =====================================================================
# STEP 2: CALCULATE LOSS (Squared Error for a single sample)
# =====================================================================
loss = (y_pred - y_true) ** 2
print(f"Current Loss: {loss:.4f}\n")

# =====================================================================
# STEP 3: BACKPROPAGATION (Calculate Gradients via Chain Rule)
# =====================================================================
# Derivative of loss with respect to prediction: dLoss/dy_pred = 2 * (y_pred - y_true)
d_loss_d_ypred = 2 * (y_pred - y_true)

# Gradients for weights: dLoss/dw = dLoss/dy_pred * dy_pred/dw (where dy_pred/dw = X)
d_loss_d_weights = d_loss_d_ypred * X

# Gradient for bias: dLoss/db = dLoss/dy_pred * dy_pred/db (where dy_pred/db = 1)
d_loss_d_bias = d_loss_d_ypred * 1

print("--- Backpropagation Gradients ---")
print(f"Gradients for Weights: {d_loss_d_weights}")
print(f"Gradient for Bias:     {d_loss_d_bias}\n")

# =====================================================================
# STEP 4: GRADIENT DESCENT (Update Parameters)
# =====================================================================
weights = weights - (learning_rate * d_loss_d_weights)
bias = bias - (learning_rate * d_loss_d_bias)

# 5. Verify the updated system state
y_pred_new = np.dot(X, weights) + bias
loss_new = (y_pred_new - y_true) ** 2

print("--- Optimization Outcome ---")
print(f"Updated Weights: {weights.round(4)}, Bias: {bias:.4f}")
print(f"New Prediction:  {y_pred_new:.4f} (Closer to true target 12.0)")
print(f"New Loss:         {loss_new:.4f} (Loss successfully minimized)")
```
