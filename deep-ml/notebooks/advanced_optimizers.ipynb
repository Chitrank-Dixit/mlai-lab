```python
import torch
import torch.nn as nn
import torch.optim as optim

# Define a mock single-layer model
model = nn.Linear(in_features=10, out_features=2)

# =====================================================================
# CONFIGURATION OPTION 1: STOCHASTIC GRADIENT DESCENT WITH MOMENTUM
# =====================================================================
# lr sets baseline speed; momentum unlocks inertial tracking
optimizer_momentum = optim.SGD(model.parameters(), lr=0.01, momentum=0.9)

# =====================================================================
# CONFIGURATION OPTION 2: RMSprop
# =====================================================================
# alpha parameter corresponds to the mathematical beta decay metric
optimizer_rmsprop = optim.RMSprop(model.parameters(), lr=0.001, alpha=0.99)

# =====================================================================
# CONFIGURATION OPTION 3: ADAM (THE HETEROGENEOUS DEFAULT)
# =====================================================================
# betas prop passes both (beta1, beta2) constraints concurrently
optimizer_adam = optim.Adam(model.parameters(), lr=0.001, betas=(0.9, 0.999))

print("--- Optimizers Initialized Seamlessly ---")
print(f"Adam Configuration Active Parameters: {optimizer_adam.defaults}")
```
