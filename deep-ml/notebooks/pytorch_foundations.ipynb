```python
import torch

# 1. Initialize input data (Fixed constants, no gradients needed)
X = torch.tensor([[1.0, 2.0], 
                  [3.0, 4.0]])

# 2. Initialize parameters (Weights and Bias require gradient tracking)
# We use float32 explicitly as the standard training precision format
w = torch.tensor([[0.5], [0.5]], dtype=torch.float32, requires_grad=True)
b = torch.tensor([0.1], dtype=torch.float32, requires_grad=True)

print("--- Initial Computational Graph Tracking ---")
print(f"Weights target gradient activation status: {w.requires_grad}")
print(f"Weights initial gradient matrix storage:    {w.grad}\n")

# 3. Executing Forward Pass (Matrix Multiplication + Bias Addition)
# y = X @ w + b
z = torch.matmul(X, w) + b

print("--- Forward Operation Evaluation ---")
print(f"Resulting output tensor z:\n{z}")
print(f"Autograd operational connection tracking function (grad_fn): {z.grad_fn}\n")

# 4. Compute a dummy loss scalar (Summing elements to yield a 1x1 value)
loss = z.sum()

# 5. Executing Backpropagation
# This triggers Autograd to trace the graph and compute all gradients
loss.backward()

print("--- Autograd Calculus Execution Complete ---")
print(f"Computed gradient matrix for Weights (dLoss/dw):\n{w.grad}")
print(f"Computed gradient scalar for Bias (dLoss/db):     {b.grad}\n")

# 6. Inference Optimization (Zero Gradient Tracking)
with torch.no_grad():
    inference_check = torch.matmul(X, w) + b
    print("--- Inference Operational State ---")
    print(f"Inference output tracking hook check: {inference_check.grad_fn} (Should be None)")
```
