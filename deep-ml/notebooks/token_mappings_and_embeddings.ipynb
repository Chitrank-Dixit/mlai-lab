```python
import torch
import torch.nn as nn

# 1. Simulate a hardcoded post-BPE vocabulary dictionary map
vocab_map = {
    "<PAD>": 0,
    "Data": 1,
    "science": 2,
    "is": 3,
    "transformative": 4
}
vocab_size = len(vocab_map)
embedding_dim = 8 # Each token will be represented by an 8-dimensional vector

print("--- Step 1: Vocabulary Map ---")
print(vocab_map)

# 2. Tokenize an incoming text string into a sequence of vocabulary indices
raw_text = ["Data", "science", "is", "transformative"]
token_ids = [vocab_map[token] for token in raw_text]
print(f"\nRaw Text Input: {raw_text}")
print(f"Mapped Token Integer IDs: {token_ids}")

# Convert the tracking array into a PyTorch batch tensor (Batch Size = 1)
input_tensor = torch.tensor([token_ids], dtype=torch.long)
print(f"Tensor Shape entering the network: {input_tensor.shape} (Batch, Seq_Len)\n")

# 3. Instantiate the PyTorch Embedding Layer
# Weights are automatically initialized randomly using a standard normal distribution
embedding_layer = nn.Embedding(num_embeddings=vocab_size, embedding_dim=embedding_dim)

# 4. Project the sequence of integer IDs into continuous vector space
with torch.no_grad():
    embedded_sequence = embedding_layer(input_tensor)

print("--- Step 2: Spatial Embedding Output ---")
print(f"Resulting Vector Tensor Shape: {embedded_sequence.shape} (Batch, Seq_Len, Embedding_Dim)")
print("\nProjected Embedding Coordinates for the token 'Data' (Row 0):")
print(embedded_sequence[0, 0].numpy().round(4))
```
