```python
import torch
import torch.nn as nn

class RegularizedNet(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super(RegularizedNet, self).__init__()
        
        # Linear Block 1 with Batch Normalization and Dropout
        self.layer1 = nn.Linear(input_dim, hidden_dim)
        # BatchNorm1d normalizes across the batch for the hidden_dim features
        self.bn1 = nn.BatchNorm1d(hidden_dim)
        self.relu = nn.ReLU()
        # Dropout drops 30% of neurons randomly during training
        self.dropout1 = nn.Dropout(p=0.3)
        
        # Final Output Layer
        self.output_layer = nn.Linear(hidden_dim, output_dim)
        
    def forward(self, x):
        # Pass through first structural block
        x = self.layer1(x)
        x = self.bn1(x)
        x = self.relu(x)
        x = self.dropout1(x)
        
        # Pass through output layer
        x = self.output_layer(x)
        return x

# Instantiate a mock model (e.g., 20 inputs, 64 hidden nodes, 1 output)
model = RegularizedNet(input_dim=20, hidden_dim=64, output_dim=1)

# Verify behavior switches between Training and Evaluation modes
mock_batch = torch.randn(5, 20)

print("--- Training Mode Active ---")
model.train() # Explicitly activates Dropout and Batch Mean/Variance calculation
train_output = model(mock_batch)
print(f"Output Tensor Shape: {train_output.shape}\n")

print("--- Inference (Evaluation) Mode Active ---")
model.eval() # Freezes BatchNorm statistics and sets Dropout rate to 0%
eval_output = model(mock_batch)
print(f"Output Tensor Shape: {eval_output.shape}")
```
