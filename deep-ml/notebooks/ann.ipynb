```python
import numpy as np

# 1. Simulate a raw feature input vector (e.g., 3 feature columns for 1 data row)
X = np.array([1.5, 2.0, -0.5])
```

```python
# 2. Define structural weights and bias for a single neuron
# We need 3 weights because we have 3 input dimensions
weights = np.array([0.4, -0.6, 0.8])
bias = -0.2
```

```python
# 3. Calculate the weighted sum (z = wx + b)
# np.dot handles the element-wise multiplication and summation simultaneously
z = np.dot(X, weights) + bias
print("--- Linear Processing Step ---")
print(f"Raw Weighted Sum (z): {z:.4f}\n")
```

```python
# 4. Define the ReLU Activation Function
def relu(val):
    return np.maximum(0, val)
```

```python
# 5. Define the Sigmoid Activation Function
def sigmoid(val):
    return 1 / (1 + np.exp(-val))
```

```python
# 6. Pass the raw output through the activation layers
output_relu = relu(z)
output_sigmoid = sigmoid(z)

print("--- Nonlinear Activation Output ---")
print(f"Neuron Output via ReLU (Hidden Layer Mode):    {output_relu:.4f}")
print(f"Neuron Output via Sigmoid (Output Prob Mode):  {output_sigmoid:.4f}")
```
