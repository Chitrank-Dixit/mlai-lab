```python
import numpy as np

# =====================================================================
# 1. MEAN SQUARED ERROR (REGRESSION)
# =====================================================================
y_true_reg = np.array([45000.0, 60000.0, 52000.0])
y_pred_reg = np.array([46000.0, 55000.0, 52200.0]) # Off by 1000, 5000, and 200

mse_loss = np.mean((y_true_reg - y_pred_reg) ** 2)
print("--- Regression Objective ---")
print(f"Computed MSE Loss: {mse_loss:.4f}\n")

# =====================================================================
# 2. BINARY CROSS-ENTROPY (BINARY CLASSIFICATION)
# =====================================================================
y_true_bin = np.array([1.0, 0.0])
y_pred_bin = np.array([0.92, 0.15]) # High confidence, accurate guesses

# Small epsilon added inside log to prevent mathematical log(0) undefined crashes
epsilon = 1e-15
y_pred_bin = np.clip(y_pred_bin, epsilon, 1 - epsilon)

bce_loss = -np.mean(y_true_bin * np.log(y_pred_bin) + (1 - y_true_bin) * np.log(1 - y_pred_bin))
print("--- Binary Classification Objective ---")
print(f"Computed BCE Loss: {bce_loss:.4f}\n")

# =====================================================================
# 3. CATEGORICAL CROSS-ENTROPY (MULTI-CLASS CLASSIFICATION)
# =====================================================================
# 3 samples, 3 possible target classes (One-Hot Encoded matrices)
y_true_multi = np.array([
    [1, 0, 0], # Sample 1 is Class 0
    [0, 1, 0], # Sample 2 is Class 1
    [0, 0, 1]  # Sample 3 is Class 2
])

# Softmax outputs (Rows sum up exactly to 1.0)
y_pred_multi = np.array([
    [0.70, 0.20, 0.10], # Decent prediction (Correct index has 0.70)
    [0.10, 0.80, 0.10], # Good prediction (Correct index has 0.80)
    [0.60, 0.30, 0.10]  # Confidently incorrect prediction (Correct index has 0.10)
])

y_pred_multi = np.clip(y_pred_multi, epsilon, 1 - epsilon)

# Sum across classes, then average across batch samples
cce_loss = -np.mean(np.sum(y_true_multi * np.log(y_pred_multi), axis=1))
print("--- Multi-Class Classification Objective ---")
print(f"Computed CCE Loss: {cce_loss:.4f}")
print("(Notice how the third confidently incorrect sample drives this loss up significantly)")
```
