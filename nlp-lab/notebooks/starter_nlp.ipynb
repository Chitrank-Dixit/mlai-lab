# NLP Lab Starter Notebook
This notebook verifies the installations of `spaCy`, `NLTK`, `LangChain`, `FAISS`, `ChromaDB` and tests basic text pipelines and MLflow connectivity.

```python
import spacy
import nltk
import langchain
import faiss
import chromadb
import mlflow

print("All libraries imported successfully!")
print(f"spaCy version: {spacy.__version__}")
print(f"NLTK version: {nltk.__version__}")
print(f"ChromaDB version: {chromadb.__version__}")
```

```python
# spaCy pipeline check
nlp = spacy.load("en_core_web_sm")
doc = nlp("Antigravity is a fantastic python library that allows us to fly.")
print("Tokens and POS tags:")
for token in doc:
    print(f"  {token.text:<15} ({token.pos_})")
```

```python
# NLTK check
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))
print(f"Number of English stopwords in NLTK: {len(stop_words)}")
```

```python
# Vector Store validation
import numpy as np
dimension = 64
nb_vectors = 10
data = np.random.random((nb_vectors, dimension)).astype('float32')

index = faiss.IndexFlatL2(dimension)
index.add(data)
print(f"FAISS index built. Total vectors: {index.ntotal}")
```

```python
# MLflow logging
mlflow.set_tracking_uri("http://mlflow:5000")
mlflow.set_experiment("nlp-starter")

with mlflow.start_run():
    mlflow.log_param("spacy_model", "en_core_web_sm")
    mlflow.log_metric("faiss_vectors_count", index.ntotal)
    print("Logged run parameters and metrics to MLflow successfully!")
```
