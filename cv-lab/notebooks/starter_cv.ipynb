# Computer Vision Lab Starter Notebook
This notebook verifies the installations of `OpenCV`, `YOLOv8`, `torchvision`, `Segment Anything`, and tests basic image manipulations and MLflow connectivity.

```python
import cv2
from ultralytics import YOLO
import torch
import torchvision
import segment_anything
import mlflow
import numpy as np
import matplotlib.pyplot as plt

print("All libraries imported successfully!")
print(f"OpenCV version: {cv2.__version__}")
print(f"Torchvision version: {torchvision.__version__}")
print(f"CUDA available for PyTorch: {torch.cuda.is_available()}")
```

```python
# OpenCV check: Create a synthetic image
image = np.zeros((300, 300, 3), dtype=np.uint8)
cv2.circle(image, (150, 150), 80, (0, 255, 0), -1) # Green circle in center

# Display using matplotlib
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("OpenCV Synthetic Circle")
plt.axis('off')
plt.show()
```

```python
# YOLOv8 check (initialize a tiny model)
try:
    model = YOLO("yolov8n.pt")
    print("YOLOv8 Nano model loaded successfully!")
    print(model.names[0:5]) # Print first 5 class names
except Exception as e:
    print(f"Could not load YOLO model (requires internet to download first time): {e}")
```

```python
# MLflow logging
mlflow.set_tracking_uri("http://mlflow:5000")
mlflow.set_experiment("cv-starter")

with mlflow.start_run():
    mlflow.log_param("opencv_version", cv2.__version__)
    mlflow.log_param("yolo_model_type", "yolov8n")
    print("Logged parameters to MLflow successfully!")
```
