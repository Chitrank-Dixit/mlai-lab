# Reinforcement Learning Lab Starter Notebook
This notebook verifies the installations of `gymnasium`, `stable-baselines3`, `ray/rllib`, `mujoco` and tests training a basic agent with MLflow logging.

```python
import gymnasium as gym
from stable_baselines3 import PPO
import ray
import mujoco
import mlflow
from pyvirtualdisplay import Display

print("All libraries imported successfully!")
print(f"Gymnasium version: {gym.__version__}")
print(f"MuJoCo version: {mujoco.__version__}")
```

```python
# Set up a virtual display to allow headless rendering
try:
    display = Display(visible=0, size=(400, 300))
    display.start()
    print("Virtual display started successfully!")
except Exception as e:
    print(f"Virtual display warning (usually fine if not rendering GUI): {e}")
```

```python
# Test Gymnasium: CartPole environment
env = gym.make("CartPole-v1")
observation, info = env.reset()
print(f"Initial observation: {observation}")
action = env.action_space.sample()
observation, reward, terminated, truncated, info = env.step(action)
print(f"Step output - reward: {reward}, terminated: {terminated}")
env.close()
```

```python
# Train stable-baselines3 PPO agent
mlflow.set_tracking_uri("http://mlflow:5000")
mlflow.set_experiment("rl-starter")

with mlflow.start_run():
    env = gym.make("CartPole-v1")
    model = PPO("MlpPolicy", env, verbose=0, learning_rate=0.003)
    
    print("Training PPO agent for 1000 steps...")
    model.learn(total_timesteps=1000)
    
    # Log parameters to MLflow
    mlflow.log_param("algorithm", "PPO")
    mlflow.log_param("total_timesteps", 1000)
    mlflow.log_param("policy", "MlpPolicy")
    
    print("Training complete! Logged run parameters to MLflow successfully.")
    env.close()
```
